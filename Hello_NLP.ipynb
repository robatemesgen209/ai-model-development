```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer

# Download required NLTK resources (once per environment)
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

# Download required NLTK resources

lemmatizer = WordNetLemmatizer()
stemmer    = PorterStemmer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # 1. Lowercase
    text = text.lower()

    # 2. Remove punctuation and special characters
    text = re.sub(r'[^a-z0-9\s]', '', text)

    # 3. Tokenize: split into individual words
    tokens = word_tokenize(text)

    # 4. Remove stopwords (common words with little meaning)
    tokens = [t for t in tokens if t not in stop_words]

    # 5. Lemmatize: reduce to base form (running -> run)
    tokens = [lemmatizer.lemmatize(t) for t in tokens]

    return tokens

# Example
text = 'The farmers are running new irrigation systems in the fields'
print(preprocess_text(text))
# Output: ['farmer', 'running', 'new', 'irrigation', 'system', 'field']


print("NLTK resources initialized successfully!")
```
